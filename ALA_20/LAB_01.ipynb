### neural network using N p array


```python
inputs=np.random.rand(10)
print(inputs)
```

```python
weights1 = np.random.rand(10)
print(weights1)
```

```python
bias1 = np.random.randint(0,10)
print(bias1)
```

```python
h1 = np.dot(inputs,weights1)+bias1
print(h1)
```

```python
weights2 = np.random.rand(10)
print(weights2)
```

```python
bias2 = np.random.randint(0,10)
print(bias2)
```

```python
h2 = np.dot(inputs,weights2)+bias2
print(h2)
```

```python
weights3 = np.random.rand(10)
print(weights3)

weights4 = np.random.rand(10)
print(weights4)
```

```python
bias3 = np.random.randint(0,10)
print(bias3)
bias4 = np.random.randint(0,10)
print(bias4)
```

```python
result = []

output1 = np.dot(h1,weights3)+ bias3
output2 = np.dot(h2,weights4)+ bias4

result.append(output1)
result.append(output2)

print(result)
```
