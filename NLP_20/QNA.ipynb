Part 1: Text Preprocessing Pipeline
1. Which field in your final pipeline was easiest to extract reliably? Why?

Word tokens were easiest to extract because whitespace-separated words are straightforward to identify in simple text.

2. Give one false positive or false negative from your work.

A false positive could occur when the word “not” is removed as a stop word, causing a negative sentence to be interpreted incorrectly.

3. Why is whitespace splitting a weak general tokenizer?

It does not handle punctuation, contractions, hyphenated words, or special characters properly.

4. Give one case where lowercasing may be harmful.

Lowercasing US to us loses the distinction between the country abbreviation and the pronoun.

5. Why can stop-word removal damage a sentiment/complaint application?

It may remove important words such as “not” or “never,” reversing the meaning of a complaint or sentiment.

6. State one precise difference between stemming and lemmatization.

Stemming removes word endings using simple rules, while lemmatization uses linguistic knowledge to return a valid dictionary base form.

7. Which preprocessing step in your pipeline is most likely to damage useful information?

Stop-word removal, because it can remove meaningful words, especially negations and contrast words.

8. What did an AI/web suggestion get wrong, oversimplify, or leave unstated?

It oversimplified tokenization by suggesting whitespace splitting without explaining how punctuation, contractions, and context affect tokens.

Part 2: Tokenization and Text Normalization
1. What is the difference between sentence tokenization and word tokenization?

Sentence tokenization splits text into sentences, while word tokenization splits sentences into individual words or tokens.

2. Give one reason .split() is not enough.

It does not reliably separate punctuation from words, such as hello, and hello.

3. Why can can't be tokenized in different ways?

Different tokenizers follow different rules. They may keep can't as one token or split it into ca and n't.

4. Why is whitespace normalization relatively low-risk?

It removes unnecessary spaces and line breaks without usually changing the meaning of the text.

5. Why is Unicode normalization useful?

It makes equivalent Unicode representations more consistent, improving text matching and comparison.

6. Give one case where lowercasing removes useful information.

Changing Apple to apple can lose the distinction between the company name and the fruit.

7. Why can stop-word removal change sentiment?

It can remove negation words such as “not,” changing a negative statement into an apparently positive one.

8. Why might but or however matter?

They indicate contrast and can introduce the main sentiment, as in “The service was slow, but the food was excellent.”

9. What is feature fragmentation?

Feature fragmentation occurs when the same underlying word or concept appears in multiple forms, such as connect, connected, and connecting, creating separate features.

10. Why can a stem be a non-dictionary word?

Stemming uses rule-based removal of word endings and does not necessarily produce a valid dictionary word.

11. Give one surprising Porter stem you observed.

For example, Porter stemming can reduce studies to studi, which is not a standard English dictionary word.

12. What is one limitation of your final pipeline?

It may lose contextual meaning through lowercasing, stop-word removal, or stemming.

13. Which preprocessing decision would you change first if the task became sentiment analysis?

I would disable stop-word removal, or at least preserve negation and contrast words such as “not,” “never,” and “but.”

Part 3: Incidence Matrices and Boolean Retrieval
1. What does 1 mean in an incidence matrix?

A 1 means that a term appears in a particular document.

2. Why does an incidence matrix contain many zeros in a large collection?

Most terms appear in only a small fraction of documents, so most term-document combinations are absent.

3. What is a postings list?

A postings list is a list of document IDs in which a particular term occurs.

4. What does df mean?

df stands for document frequency: the number of documents containing a particular term.

5. How is an incidence row related to a postings list?

An incidence row records term presence across all documents, while a postings list stores only the IDs of documents where the term is present.

6. How is Boolean AND performed using postings lists?

By intersecting the two postings lists and keeping only the document IDs present in both lists.

7. Why must documents and queries use compatible preprocessing?

Otherwise, the same word may be represented differently in the document index and the query, causing relevant documents to be missed.

8. Give one limitation of exact Boolean retrieval.

It returns documents that exactly satisfy the Boolean query but does not rank them by relevance or handle synonyms automatically.

2. What is the Markov assumption in a bigram model?

It assumes that the probability of the current word depends only on the immediately preceding word.

3. Why are sentence boundary symbols useful?

They help the model identify where sentences begin and end, allowing it to learn sentence-start and sentence-ending patterns.

4. Why can a trigram model be more data-hungry than a bigram model?

It considers two previous words, creating more possible word combinations that require training examples to estimate their probabilities reliably.

5. Why is perplexity usually preferred over raw sentence probability when comparing sequences of different lengths?

Perplexity normalizes probability by sequence length, making comparisons between sentences of different lengths more meaningful.

6. Why can an unsmoothed N-gram model assign infinite perplexity to a perfectly grammatical sentence?

If any N-gram in the sentence has zero probability because it was never observed in training, the sentence probability becomes zero, resulting in infinite perplexity.
