### Problem 1 : A fair coin is tossed 100 times


```python
import numpy as np 
import random

toss = [random.choice(["Heads", "Tails"]) for _ in range(100)]
print(toss)

#number of Heads 
heads = toss.count("Heads")
print("number of Heads: ",heads)

#number of Tails 
tails = toss.count("Tails")
print("number of tails",tails)

#probability of Heads 
print("Probability of Heads: ",(heads)/len(toss))

#probability of tails 
print("Probability of Tails: ",(tails)/len(toss))
```

### Problem 2: Check the effect of the number of trails 

```python
import numpy as np 
import random

trails=[10,100,1000,10000]

for i in range(len(trails)):
    print(f"\nTrials size : {trails[i]}")
    toss = [random.choice(["Heads", "Tails"]) for _ in range(trails[i])]
    print(toss)

    #number of Heads 
    heads = toss.count("Heads")
    print("number of Heads: ",heads)

    #number of Tails 
    tails = toss.count("Tails")
    print("number of tails",tails)

    #probability of Heads 
    print("Probability of Heads: ",(heads)/len(toss))

    #probability of tails 
    print("Probability of Tails: ",(tails)/len(toss))
    
```

### Problem 3: Rolling a die 6000 times

```python
outcomes = np.random.randint(1,7,size=6000)

for i in range(1,7):
    count= np.sum(outcomes == i)
    probability = count/len(outcomes)

    print(
        'Outcome',i,
        'Count',count,
        'Experimental Probability',probability
    )
```

### Problem 4 : Addition rule for disjoint events

```python
results= np.random.randint(1,7,size=10000)

a_count= np.count_nonzero(results == 1)
b_count= np.count_nonzero(results == 6)

A_and_B= a_count+b_count

a_prob = a_count/len(results)
b_prob = b_count/len(results)

print('P(A): ',a_prob)
print('P(B): ',b_prob)

A_and_B_prob =  A_and_B/len(results)

print('P (A U B) :  ',A_and_B_prob)

Add_prob =  a_prob + b_prob 
print(f'P(A) + P(B) : {Add_prob:.4f}')

if A_and_B_prob == Add_prob:
    print("Hence proved it's a Disjoint event")
else:
    print("It's a joint event ")
```

### Problem 5: Addition Rule for overlapping events 

```python
results= np.random.randint(1,7,size=10000)

A = np.isin(results, [2, 4, 6])
B = np.isin(results, [4, 5, 6])

a_count= np.count_nonzero(A)
b_count= np.count_nonzero(B)

intersection_count= np.count_nonzero(A & B)
union_count = np.count_nonzero (A | B)



a_prob = a_count/len(results)
b_prob = b_count/len(results)

print('P(A): ',a_prob)
print('P(B): ',b_prob)

intersection_prob = intersection_count / len(results)
union_prob =  union_count/len(results)

print('P(A n B) : ', intersection_prob)

print('P (A U B) :  ',union_prob)


if intersection_count == 0:
    print("Hence proved it's a Disjoint event")
else:
    print("It's a joint event ")
```

### Problem 6: Multiplication rule for Independent events 

```python
n=10000

coin = np.random.choice(['H','T'],size=n)
dice=np.random.randint(1,7,size=n)

A = coin== 'H'
B = dice%2 == 0

prob_A = np.mean(A)
prob_B = np.mean(B)
prob_intersection = np.mean(A & B) 

print('P(A) : ',prob_A)
print('P(B) : ',prob_B)
print('P(A n B) : ',prob_intersection)
print('P(A) x P(B) : ',prob_A * prob_B)
```
