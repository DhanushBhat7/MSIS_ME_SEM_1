```python
import numpy as np 
import matplotlib.pyplot as plt

n_people=np.arange(1,50)
prob_shared=[]
prob_no_shared=1.0

for i in n_people:
    prob_no_shared*=(365-(i-1))/365
    prob_shared.append(1-prob_no_shared)

print(prob_shared)

plt.figure(figsize=(8,5))
plt.scatter(n_people,prob_shared, color="blue", s=30, zorder=3, label="Probability per Group Size")
plt.xlabel("Number of People")
plt.ylabel("Probability of Shared Birthday")
plt.title("The Birthday Paradox")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Theoretical Calculation
n_people = np.arange(1, 100)
prob_shared = []

p_no_shared = 1.0
for n in n_people:
    p_no_shared *= (365 - (n - 1)) / 365
    prob_shared.append(1 - p_no_shared)

print(prob_shared)

# Plotting the theoretical curve
plt.figure(figsize=(8, 5))
plt.scatter(n_people, prob_shared, color="blue", s=30, zorder=3, label="Probability per Group Size")
plt.xlabel("Number of People")
plt.ylabel("Probability of Shared Birthday")
plt.title("The Birthday Paradox")
plt.legend()
plt.grid(True)
plt.show()
```
